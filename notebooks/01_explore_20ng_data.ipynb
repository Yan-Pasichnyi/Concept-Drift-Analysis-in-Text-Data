# 20 Newsgroups Stream Exploration

In this notebook, I load the raw 20 Newsgroups dataset (with headers, footers, and quotes removed), encode the texts using the frozen `BAAI/bge-small-en-v1.5` model, and check the structure of the abrupt and gradual drift streams.

## Setup and Imports

```python
import sys
import os
from collections import Counter

sys.path.append(os.path.abspath('../src'))
from data_loader import load_raw_20ng, get_canonical_embeddings, stream_generator
```

## Raw Texts and Embeddings

```python
texts, topics = load_raw_20ng()
embeddings, _ = get_canonical_embeddings()

print("Total documents:", len(texts))
print("Embedding vector length:", len(embeddings[0]))
print("Topics distribution:", Counter(topics))
print("First text sample:", texts[0][:200])
```

## Check Abrupt and Gradual Streams

```python
first_abr_x, first_abr_y = next(stream_generator(mode='abrupt'))
first_grad_x, first_grad_y = next(stream_generator(mode='gradual'))

print("Abrupt stream features:", len(first_abr_x), "| First label:", first_abr_y)
print("Gradual stream features:", len(first_grad_x), "| First label:", first_grad_y)
```

## Summary

- The cleaned dataset contains **15,518** documents across 4 super-topics: `comp` (4,750), `sci` (3,822), `rec` (3,797), and `talk` (3,149).
- Each document is encoded independently into a **384-dimensional** embedding vector.
- Both `abrupt` and `gradual` streams yield feature dictionaries with 384 features and binary labels (`0` or `1`), ready for prequential evaluation in `river`.
